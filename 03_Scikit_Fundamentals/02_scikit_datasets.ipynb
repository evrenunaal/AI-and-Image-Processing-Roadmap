```python
#MAKE FUNK
from sklearn.datasets import make_blobs,make_moons
#stickit kutuphanesi uzerinde bulunan datasets modulunden make_bolbs kume olusturucusunu cagirir
#veya moons ile beraber
```

```python
X,y = make_blobs(n_samples=5000,centers=2)
#X:Makinenin ogrenirken bakacagi ip uclari dev bir tablo gibi dusun (OZELLIK MATRISI)
#y:Cevap anahtari gibi kullanilir X teki her bir satirin hangi sinifa ait oldugunu tutar
#n_samples=: make_bolbs ile beraber kac satirlik bir veri uretilecegine karar veririz
#centers=:VErileri 2 farkli gruba boler yani y degiskenini sadece 1 ve 0 lardan olusturur
#n_features=: Her bir verinin 5 farkli ozelliginin olmasini saglar
```

```python
X
```

```python
y
```

```python
plt.scatter(X[:,0],X[:,1], c=y) #Her calistirmada farkli bir veri seti olusturdugundan kaynakli grafiklerde degislik olacaktir
```

```python
X,y = make_moons(noise=0.01,random_state=0)
plt.scatter(X[:,0],X[:,1], c=y)
#noise ile beraber duzen saglanir
#random_state=0 ile beraber veri seti her zaman ayni cikacak sekilde olur eger her defasinda ayni tutarli veri setini istiyorsak o zaman 42 yazilmali  
```
